```python
def gtf_to_bed(filepath, bed_path):
    
    f = gzip.open(filepath, 'rt')
    b = open(bed_path, 'w')

    for line in f:
        if line.startswith('#'):
            continue

        cols = line.split('\t')
        if cols[2] != 'gene':
            continue
        if cols[8].split('gene_type "')[1].split('"')[0] != 'protein_coding':
            continue

        chrom = cols[0]
        start = cols[3]
        end = cols[4]
        strand = cols[6]
        
    
        name = cols[8].split('gene_name "')[1].split('"')[0]
        genecode = cols[8].split('gene_id "')[1].split('"')[0]
        gene_name = name + '_' + genecode

        bed_line = f"{chrom}\t{start}\t{end}\t{gene_name}\t{strand}\n"
        b.write(bed_line)
```

```python
bed_path = os.path.join(wd, 'output.bed')
gtf_to_bed(filepath, bed_path)
```

```python
def gtf_bed_promoters(filepath, bed_path):
    f = gzip.open(filepath, 'rt')
    b = open(bed_path, 'w')

    for line in f:
        if line.startswith('#'):
            continue

        cols = line.split('\t')
        if cols[2] != 'gene':
            continue
        if cols[8].split('gene_type "')[1].split('"')[0] != 'protein_coding':
            continue
            
        chrom = cols[0]
        strand = cols[6]

        if strand == '+':
            start = int(cols[3]) - 500
            end = int(cols[3]) + 100

        else:
            start = int(cols[4]) + 500
            end = int(cols[4]) - 100
            
        name = cols[8].split('gene_name "')[1].split('"')[0]
        genecode = cols[8].split('gene_id "')[1].split('"')[0]
        gene_name = name + '_' + genecode

        bed_line = f"{chrom}\t{start}\t{end}\t{gene_name}\t{strand}\n"
        b.write(bed_line)
```

```python
bed_pro = os.path.join(wd, 'promoters.bed')
gtf_bed_promoters(filepath, bed_pro)
```
