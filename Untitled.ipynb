```python
#1.1/1.3
def DNA_sequence(seq_length):
    output = ''
    for x in range(seq_length):
        add = np.random.choice(['A', 'T','C','G'], p= [0.25,0.25,0.25,0.25])
        output += add
    print(output)
    return output
```

```python
#1.2
def FASTA_formatting (filepath, nametag, sequence):
    DNA = open(filepath, 'a')
    DNA.write('>' + nametag + '\n')
    num_chunks = len(sequence)//80
    for i in range(0, num_chunks+1):
        chunk = sequence[i*80 : (i+1)*80]
        DNA.write(chunk + '\n')
    DNA.close()
```

```python
#1.4
def FASTA_finding_seq(filepath, nametag):
    Seq = ''
    DNA = open(filepath, 'r')
    found_tag = False
    for line in DNA:
        line = line.strip()
        if line.startswith('>'):
            if line[1:] == nametag:
                found_tag = True
                Seq = ''
            else:
                found_tag = False
        elif found_tag:
            Seq += line
    return Seq
```

```python
#1.5
import random

nucleotides = ["A", "T", "C", "G"]

cg_total = 0.42
c_weight = random.uniform(0, cg_total)
g_weight = cg_total - c_weight

# Dynamische Gewichtung für A und T
at_total = 1 - cg_total  # 58% für A und T
a_weight = random.uniform(0, at_total)  # Zufällige Verteilung für A
t_weight = at_total - a_weight  # Der Rest geht an T

weights = [a_weight, t_weight, c_weight, g_weight]

# Erzeugen der Sequenz
seq = ''.join(random.choices(nucleotides, weights=weights, k=100))

print("Gewichtungen:", weights)
print("Sequenz:", seq)
```

```python
#2.1
def pos_pattern (DNA_seq, pattern):
    positions = ''
    for i in range(0, len(DNA_seq)):
        if DNA_seq[i:i + len(pattern)] == pattern:
            positions += str(i) + ' '
    return positions
        
```

```python
#2.2
import re
def pos_pat_reg (DNA_seq, pattern):
    pos_pat = [match.start() for match in re.finditer(pattern, DNA_seq)]
    return pos_pat
```

```python
#2.3
seq = DNA_sequence(10000)
BseBI = 'CC[A,T]GG'
HaeII = '[A,G]GCGC[C,T]'
KoxII = 'G[A,G]GC[C,T]C'
```

```python
def pos_RS (DNA_seq):
    pos_BseBI = 'Position BseBI: ' + str([match.start() for match in re.finditer(BseBI, DNA_seq)]) + '\n'
    pos_HaeII = 'Position HaeII: ' + str([match.start() for match in re.finditer(HaeII, DNA_seq)]) + '\n'
    pos_KoxII = 'Position KoxII: ' + str([match.start() for match in re.finditer(KoxII, DNA_seq)]) + '\n'
    pos_all = pos_BseBI + pos_HaeII + pos_KoxII
    return pos_all
```

```python
def frag_len(DNA_seq):
    sites = [0]
    for match in re.finditer(BseBI, DNA_seq):
        sites.append(match.start())
    for match in re.finditer(HaeII, DNA_seq):
        sites.append(match.start())
    for match in re.finditer(KoxII, DNA_seq):
        sites.append(match.start())
    sites.append(len(DNA_seq))
    sites.sort()
    print(sites)
    
    diff= []
    for i in range(0, len(sites)-1):
        diff.append(sites[i+1] - sites[i])

    diff.sort()
        
    freq_dict = {}
    for length in diff:
        if length in freq_dict:
            freq_dict[length] += 1
        else:
            freq_dict[length] = 1

    rel_freq_dict = {}
    for length in freq_dict:
        abs_freq = freq_dict[length]
        rel_freq_dict[length] = abs_freq/len(diff)

    print(f"{'Length':<10}{'Absolute Frequency':<20}{'Relative Frequency'}")  # Header
    for length in freq_dict:  # Durch alle Fragmentlängen iterieren
        abs_freq = freq_dict[length]
        rel_freq = rel_freq_dict[length]
        print(f"{length:<10}{abs_freq:<20}{rel_freq:.4f}")  # Jede Zeile der Tabelle

    


    
    return
        
```
