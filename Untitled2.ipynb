```python
def bed_to_dict(bed_file):
    dictionary = {}
    b = open(bed_file, 'r')
    for line in b:
        chrom, start, end, name, strand = line.strip().split('\t')
       
        if chrom not in dictionary:
                dictionary[chrom] = {}

        dictionary[chrom][int(start)] = (chrom, start, end, name, strand)
    return dictionary
```

```python
 dict = bed_to_dict(bed_path)
```

```python
dict_chr1 = dict['chr1']
keys = list(dict_chr1.keys())

#keys.list()
sorted(keys)
print(type(keys))
print(keys)
```

```python
def overlap_gene_finder(chrom, start, end):
    dict_chrom = dict[chrom]
    keys = list(dict_chrom.keys())

    sorted(keys)
    overlap = []
    for k in keys:
        if start <= k <= end:
            overlap.append(k)

        genes = []
        for key in overlap:
            genes.append(dict[chrom][key])

    return genes
```
